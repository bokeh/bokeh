# Automatic MIME display

A Bokeh object left as the final cell expression displays automatically as a static snapshot. No initialization call or separate Jupyter extension installation is needed.

```python
from bokeh.plotting import figure
```

```python
p = figure(title="Automatic static snapshot")
p.line([1, 2, 3], [3, 1, 2])
p
```

Use `show(p)` when later Python property changes should update every open Jupyter view automatically.
