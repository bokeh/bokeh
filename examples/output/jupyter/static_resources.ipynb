# Static notebook output and shared resources

Bokeh 4.0 detects the notebook automatically. The first output stores each INLINE BokehJS artifact once; later outputs refer to it.

```python
from IPython.display import display
from bokeh.plotting import figure
from bokeh.settings import settings
settings.resources = "inline"
```

```python
p = figure(title="First static snapshot")
p.line([1, 2, 3], [3, 1, 2])
display(p)
```

```python
q = figure(title="Shares the INLINE BokehJS owner")
q.scatter([1, 2, 3], [2, 3, 1], size=14)
display(q)
```

Automatic/final-expression snapshots use `settings.resources` (CDN by default). Explicit connected output can select resources per view with `show(q, resources=CDN)`, or `Resources(mode="server", root_url=...)` for a local static resource server.
