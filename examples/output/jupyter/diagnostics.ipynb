# Notebook diagnostics

The Bokeh renderer reports failures in the output with a stable code, a recovery action, redacted technical details, and a **Copy report** button.

Common codes:

- `RESOURCE_RECORD_MISSING`: re-run the display in the bundled renderer so it can recover a deleted owner over a live-kernel comm; with the executable fallback, restart the kernel and run the notebook again.
- `RESOURCE_LOAD_FAILED`: check CDN/server URLs or use `INLINE`; if the diagnostic reports a timeout, reload the notebook page before retrying or changing modes (a kernel restart alone is insufficient).
- `BOKEH_VERSION_MISMATCH`: restart the kernel and reload the notebook.
- `APPLICATION_RENDER_FAILED`: ensure the ASGI app is running and `jupyter-server-proxy` is installed for remote Jupyter; other remote hosts may need a `notebook_url` override on `serve(...)`.

```python
from bokeh.io import show
from bokeh.plotting import figure

p = figure(title="Healthy output")
p.line([1, 2, 3], [1, 3, 2])
show(p)
```

The next cell intentionally points newly required table/widget resources at a closed local port. Its expected output is a `RESOURCE_LOAD_FAILED` panel. Expand **Technical details** and try **Copy report**; then switch the call to `resources=recovery_resources` to recover.

```python
from bokeh.models import ColumnDataSource, DataTable, TableColumn
from bokeh.resources import INLINE, Resources

table = DataTable(
    source=ColumnDataSource(data={"x": [1, 2]}),
    columns=[TableColumn(field="x", title="x")],
)
recovery_resources = INLINE
unreachable = Resources(mode="server", root_url="http://127.0.0.1:1/")
show(table, resources=unreachable)
```
