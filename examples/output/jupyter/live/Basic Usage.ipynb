```python
from bokeh.io import show
from bokeh.layouts import row
from bokeh.plotting import figure
```

```python
opts = dict(width=250, height=250, min_border=0)
```

```python
p1 = figure(**opts)
r1 = p1.scatter([1,2,3], [4,5,6], size=20)

p2 = figure(**opts)
r2 = p2.scatter([1,2,3], [4,5,6], size=20)

# create a live output handle
t = show(row(p1, p2))
```

```python
# the live handle repr shows which output cell it manages
t
```

```python
# model changes synchronize automatically
r1.glyph.fill_color = "white"
```

```python
r2.glyph.fill_color = "pink"
```

```python
p3 = figure(**opts)
r3 = p3.scatter([1,2,3], [4,5,6], size=20)

# create a live output handle
t2 = show(p3)
```

```python
# show which output cell t2 manages
t2
```

```python
# this updates the output managed by t2
r3.glyph.fill_color = "orange"
```

```python
# this updates the first live output
r1.glyph.fill_color = "orange"
```

```python
# create a live output handle
t3 = show(p2)
```

```python
# p2 now appears in two live outputs, so both views update
r2.glyph.fill_color = "red"
```

```python
# hold batches related changes into one update for this handle
with t3:
    r2.glyph.fill_color = "blue"
    r2.glyph.line_color = "navy"
```
