# Execution order and Python identity

`show(p)` automatically synchronizes attached Bokeh models. Re-running the HoverTool assignment below still creates a new unattached object; no renderer can infer that the Python name should replace the tool already in `p`. Connected updates are browser-tested with the bundled renderer in JupyterLab 4 and Notebook 7.

```python
from bokeh.io import show
from bokeh.models import HoverTool
from bokeh.plotting import figure
```

```python
p = figure()
p.scatter([1, 2, 3], [3, 1, 2], size=15)
```

```python
hover = HoverTool(tooltips=[("x", "$x"), ("y", "$y")])
```

```python
p.add_tools(hover)
handle = show(p)
```

```python
hover.tooltips = [("index", "$index")]
```

Re-running only the `hover = HoverTool(...)` cell does not update the attached tool. Re-running `p.add_tools(hover)` would add a duplicate. Rebuild and re-show the plot, or select and update the attached tool explicitly.
